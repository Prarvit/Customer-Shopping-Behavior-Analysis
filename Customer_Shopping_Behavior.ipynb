```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform( lambda x: x.fillna(x.median()))
# filling the null values with median of that category.
```

```python
df.isnull().sum()
```

```python
# The snake_casing(lowercase and space = '_') the columns as it is hard to write the upper, lowercase and special charracters
df.columns= df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns = { 'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
df.columns # looks neat and consistent
```

```python
# Create different age_groups
# qcut = Quantile based Discretization, here dividing the total age group in 4 parts (25%) and putting labels on those 4 sub-categories)
labels = [ 'Young Adults' , ' Adult', 'Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'], q =4, labels = labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
# Text to numbers of frequency of purchase to  make nalysis easiler to understand

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly' : 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months' : 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
